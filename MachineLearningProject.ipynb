```python
def Prepare_data_for_models(train, test, use_polynomial=False):
    features = get_features()

    X_train, y_train, poly_transformer = clean_data(train, features, use_polynomial)

    X_test, test_IDs = clean_test_data(test, features, poly_transformer)

    return X_train, y_train, X_test, test_IDs, features

def clean_data(train, features, use_polynomial=False, degree=2):
    df = CleanDataFunctions(train)
    df = CalculateParameters(df)

    X = df[features].values
    y = df['Target Pressure (bar)'].values

    # Initialize and apply PolynomialFeatures only if requested
    if use_polynomial:
        poly = PolynomialFeatures(degree=degree, include_bias=False)
        X = poly.fit_transform(X)
        return X, y, poly
    else:
        # Return identity polynomial transformer when not using polynomial features
        poly = PolynomialFeatures(degree=1, include_bias=False)
        poly.fit(X)  # Fitting with original features (no transformation)
        return X, y, poly

def clean_test_data(test, features, poly_transformer):
    test_cleaned = standardize_status_values(test)
    test_cleaned = CalculateParameters(test_cleaned)

    test_X = test_cleaned[features].values
    IDs = test_cleaned['ID'].values

    # Apply the same polynomial transformation to the test data
    X_test_poly = poly_transformer.transform(test_X)

    return X_test_poly, IDs
```

```python
def CleanDataFunctions(df):

  noNulls = Remove_Null_Values(df)
  noOutliers = Remove_Outliers(noNulls)
  noDuplicates = remove_duplicates(noOutliers)
  standardisedData = standardize_status_values(noDuplicates)

  return standardisedData

def Remove_Null_Values(df):
    initial_rows = df.shape[0]

    # Drop rows with any null values
    noNulls = df.dropna()

    final_rows = noNulls.shape[0]
    null_rows_removed = initial_rows - final_rows

    # Calculate the total number of null values before removal
    total_nulls = df.isnull().sum().sum()

    #print(f"Initial number of rows: {initial_rows}")
    #print(f"Total null values: {total_nulls}")
    #print(f"Number of rows removed: {null_rows_removed}")
    #print(f"Number of rows after removal: {final_rows}")

    return noNulls

def Remove_Outliers(df):
    initial_rows = df.shape[0]

    # Calculate the Z-scores for each column in the DataFrame
    z_scores = stats.zscore(df.select_dtypes(include=[np.number]))

    # A boolean mask for rows with any Z-scores greater than 3 or less than -3
    mask = (np.abs(z_scores) < 3).all(axis=1)

    # Apply the mask to select only the rows without outliers
    df_no_outliers = df[mask]

    final_rows = df_no_outliers.shape[0]
    outliers_removed = initial_rows - final_rows

    #print(f"Initial number of rows: {initial_rows}")
    #print(f"Number of outliers removed: {outliers_removed}")
    #print(f"Number of rows after removal: {final_rows}")

    return df_no_outliers

def remove_duplicates(df, subset=None, keep='first'):

    initial_rows = df.shape[0]

    # Removing duplicates
    df_no_duplicates = df.drop_duplicates(subset=subset, keep=keep)

    final_rows = df_no_duplicates.shape[0]
    duplicates_removed = initial_rows - final_rows

    #print(f"Initial number of rows: {initial_rows}")
    #print(f"Number of duplicate rows removed: {duplicates_removed}")
    #print(f"Number of rows after removal: {final_rows}")

    return df_no_duplicates

def standardize_status_values(df):
    df_copy = df.copy()

    status_mapping = {
        'Subcooled': 0,
        'subcooled': 0,
        'Subcool': 0,
        'Subcoled': 0,
        'Superheated': 1,
        'superheated': 1,
        'Superheat': 1,
        'Saperheated': 1
    }

    # Using direct column assignment to avoid deprecation warnings
    df_copy['Status'] = df_copy['Status'].map(status_mapping).fillna(-1)

    return df_copy

def scale_And_Normalize_Data(df):
    # Initialize the StandardScaler and MinMaxScaler
    standard_scaler = StandardScaler()
    min_max_scaler = MinMaxScaler()

    # Standardizing the data
    standardized_data = standard_scaler.fit_transform(df)
    df_standardized = pd.DataFrame(standardized_data, columns=df.columns)

    # Normalizing the data
    normalized_data = min_max_scaler.fit_transform(df)
    df_normalized = pd.DataFrame(normalized_data, columns=df.columns)

    return df_standardized, df_normalized

def CalculateParameters(df):

  df['Tank Volume (m³)'] = calculate_Tank_Volume(df)
  df['Obstacle Volume (m³)'] = calculate_Obstacle_Volume(df)
  df['Sensor Distance (m)'] = calculate_Vector_Distance(df)

  return df

def calculate_Tank_Volume(df):
    tank_volume = df['Tank Width (m)'] * df['Tank Length (m)'] * df['Tank Height (m)']
    return tank_volume

def calculate_Obstacle_Volume(df):
    obstacle_volume = df['Obstacle Width (m)'] * df['Obstacle Height (m)'] * df['Obstacle Thickness (m)']
    return obstacle_volume

def calculate_Vector_Distance(df):
    # Calculate the  distance
    distances = np.sqrt((df['Sensor Position x'] ** 2) +
                        (df['Sensor Position y'] ** 2) +
                        (df['Sensor Position z'] **  2)
                        )
    return distances
    """
def calculate_Vector_Distance_Improved(df):

    # Calculate the new origin based on the center of the tank
    new_origin_x = df['Tank Width (m)'] / 2
    new_origin_y = df['Tank Height (m)'] / 2
    new_origin_z = df['Tank Length (m)'] / 2

    # Adjust sensor coordinates to the new origin
    adjusted_x = df['Sensor Position x'] - new_origin_x
    adjusted_y = df['Sensor Position y'] - new_origin_y
    adjusted_z = df['Sensor Position z'] - new_origin_z

    # Calculate the distance from the new origin
    distances = np.sqrt(adjusted_x ** 2 + adjusted_y ** 2 + adjusted_z ** 2)
    return distances
    """
```

```python
def create_linear_regression_model(X_train, y_train, X_test, test_IDs, features):
    # Initialize and scale features
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Create and train the model
    model = LinearRegressionModel()
    model.fit(X_scaled, y_train)

    return model

class LinearRegressionModel:
    def __init__(self):
        self.w = None
        self.b = 0.0
        self.J_history = []

    def fit(self, X, y, alpha=0.001, num_iters=10000):
        n, d = X.shape
        self.w = np.zeros(d)
        self.b = 0.0

        for i in range(num_iters):
            predictions = np.dot(X, self.w) + self.b
            errors = predictions - y

            dj_dw = (1 / n) * np.dot(X.T, errors)
            dj_db = (1 / n) * np.sum(errors)

            self.w -= alpha * dj_dw
            self.b -= alpha * dj_db

            if i % 100 == 0:
                cost = (1 / (2 * n)) * np.sum((errors) ** 2)
                self.J_history.append(cost)

    def predict(self, X):
        return np.dot(X, self.w) + self.b
```

```python
"""
No longer used functions:
"""

def print_column_ranges(df):
  for column in df.columns:
      # Check if the column is numeric
      if pd.api.types.is_numeric_dtype(df[column]):
          min_val = df[column].min()
          max_val = df[column].max()
          # Calculate the range of values (max - min)
          value_range = max_val - min_val
          print(f"Column: {column}  Min: {min_val}, Max: {max_val}, Range: {value_range}\n")
      else:
          # For non-numeric columns, indicate that range cannot be calculated
          print(f"Column: {column} is non-numeric, so range is not applicable.\n")


def filter_data_by_range(df, column, min_val, max_val):
    initial_count = df.shape[0]
    filtered_df = df[(df[column] >= min_val) & (df[column] <= max_val)]
    final_count = filtered_df.shape[0]
    outside_range_count = initial_count - final_count

    print(f"For column '{column}': Initial count = {initial_count}, Values outside the range ({min_val}, {max_val}) = {outside_range_count}, Final count = {final_count}")

    # Return both the filtered DataFrame and the count of rows removed
    return filtered_df, outside_range_count

def create_lightgbm_model(X_train, y_train, X_test, test_IDs, features, n_iter=100, cv=5):
    # Optional: Standardize features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    # Parameter distributions for RandomizedSearchCV
    param_distributions = {
        'num_leaves': np.arange(20, 61, 10),
        'learning_rate': [0.001, 0.01, 0.05, 0.1],
        'feature_fraction': [0.8, 0.9, 1.0],
        'bagging_fraction': [0.8, 0.9, 1.0],
        'bagging_freq': [5, 7, 9],
        'n_estimators': [100, 200, 500]
    }

    # Create a LightGBM regressor object
    lgbm = LGBMRegressor(objective='regression', metric='rmse', random_state=42)

    # Setup RandomizedSearchCV
    randomized_search = RandomizedSearchCV(
        estimator=lgbm,
        param_distributions=param_distributions,
        n_iter=n_iter,  # Number of parameter settings that are sampled
        cv=cv,  # Cross-validation folding
        scoring='neg_root_mean_squared_error',  # Scoring method
        verbose=1,
        n_jobs=-1,  # Use all available cores
        random_state=42
    )

    # Perform the hyperparameter search
    randomized_search.fit(X_train_scaled, y_train)

    # Extract the best estimator
    best_model = randomized_search.best_estimator_

    # Make predictions using the best model
    test_predictions = best_model.predict(X_test_scaled)

    # Optionally create a submission or handle predictions as needed
    create_submission(test_predictions, test_IDs)

    # Output best parameters and score
    print("Best parameters found: ", randomized_search.best_params_)
    print("Best score found: ", -randomized_search.best_score_)  # Note negation for RMSE

    return best_model

def create_svm_model(X_train, y_train, X_test, test_IDs, features, C=1.0, epsilon=0.1):

    model = SVR(C=C, epsilon=epsilon, kernel='rbf')
    model.fit(X_train, y_train)

    return model

def create_xgboost_model(X_train, y_train, X_test, test_IDs, features):
    # Define the XGBoost parameters
    params = {
        'objective': 'reg:squarederror',
        'max_depth': 6,
        'min_child_weight': 4,
        'subsample': 0.8,
        'colsample_bytree': 0.9,
        'learning_rate': 0.05,
        'n_estimators': 1000
    }

    # Setup KFold
    n_splits = 5
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

    # List to store each fold's RMSE
    rmse_scores = []

    # Perform K-fold cross-validation
    for train_index, val_index in kf.split(X_train):
        X_train_fold, X_val_fold = X_train[train_index], X_train[val_index]
        y_train_fold, y_val_fold = y_train[train_index], y_train[val_index]

        # Convert data to DMatrix form
        dtrain_fold = xgb.DMatrix(X_train_fold, label=y_train_fold)
        dval_fold = xgb.DMatrix(X_val_fold, label=y_val_fold)

        # Train the model
        model = xgb.train(params, dtrain_fold, num_boost_round=1000, evals=[(dval_fold, 'eval')],
                          early_stopping_rounds=10, verbose_eval=False)

        # Predict on validation fold
        preds = model.predict(dval_fold)

        # Calculate RMSE for the current fold
        rmse = np.sqrt(mean_squared_error(y_val_fold, preds))
        rmse_scores.append(rmse)

    # Output average RMSE across all folds
    average_rmse = np.mean(rmse_scores)
    print(f"Average RMSE across {n_splits} folds: {average_rmse}")

    # Retrain model on the entire training set
    dtrain = xgb.DMatrix(X_train, label=y_train)
    dtest = xgb.DMatrix(X_test)
    final_model = xgb.train(params, dtrain, num_boost_round=model.best_iteration)

    return final_model
```
